Given array A, return the smallest positive integer (>0) not present.

A = [1, 2, 3] → 4
A = [-1, -3] → 1

```python
def smallest_positive_integer(A: list[int]) -> int:
    """Returns the smallest positive integer."""

    if not A:
        return 1
    
    unique_A = set(A)
    max_x = 0

    for d in unique_A:
        if d > max_x:
            max_x = d
    
    if max_x <= 0:
        return 1
    else:
        return max_x + 1

## Test
A = [-1, -3]
answer = smallest_positive_integer(A)
print(answer)
```

"{[()()]}" → valid
"([)()]" → invalid

```python
def brackets_counter(s: str) -> bool:
    bracket_dict = {'(': ')', '[': ']', '{': '}'}
    stack = []
    for i, s_ in enumerate(s):
        if s_ in bracket_dict:
            stack.append(s_)
        else:
            if len(stack) == 0:
                return False
            
            last = stack.pop()
            if s_ != bracket_dict[last]:
                return False
    return len(stack) == 0

given = "((("
answer = brackets_counter(given)
print(answer)
```

```python
def tape_equilibrium(a: list[int]) -> int:
    left_sum = 0
    total = sum(a)
    min_diff = float('inf')

    for i in range(len(a) - 1):
        left_sum += a[i]
        right_sum = total - left_sum
        min_diff = min(min_diff, abs(left_sum - right_sum))

    return min_diff

A = [3, 1, 2, 4, 3]
answer = tape_equilibrium(A)
print(answer)
```

```python
def passing_cars(a: list[int]) -> int:
    east_count = 0
    pairs = 0

    for car in a:
        if car == 0:
            east_count += 1
        else:
            pairs += east_count

    return pairs

A = [0, 1, 0, 1, 1]
answer = passing_cars(A)
print(answer)
```
